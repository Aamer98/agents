```python
import asyncio
import time
import gradio as gr
```

```python
async def sleep_and_print(i):
    await asyncio.sleep(i)
    print(f"Slept for {i} seconds")
```

```python
def blocking_sleep_and_print(i):
    time.sleep(i)
    print(f"Slept for {i} seconds")
```

```python
await sleep_and_print(2)
```

```python
async def gather_results():
    results = await asyncio.gather(
        sleep_and_print(1),
        sleep_and_print(2),
        sleep_and_print(3)
    )
    return results
```

```python
await gather_results()
```

```python
def gather_blocking_results():
    blocking_sleep_and_print(1)
    blocking_sleep_and_print(2)
    blocking_sleep_and_print(3)
    
```

```python
gather_blocking_results()
```

```python
import gradio as gr
import asyncio

async def slow_response(name):
    await asyncio.sleep(2)
    return f"Hello, {name}! (after waiting)"


def blocking_response(name):
    time.sleep(2)
    return f"Hello, {name}! (after waiting)"

gr.Interface(fn=blocking_response, inputs="text", outputs="text").launch()
```
